```python
# Importaciones mínimas: usar módulo externo para generación de datos
import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parent))

from src.utils.knapack_parser import knapack_parser
from src.utils.generate_eo_data import generate_data

# Solo numpy si quieres ajustar rangos dinámicamente
import numpy as np
```

```python
# Rutas de instancias
small_instances_paths = [
    "smallcoeff_pisinger\\knapPI_1_10000_10000.csv",
    "smallcoeff_pisinger\\knapPI_2_10000_10000.csv",
]
large_instances_paths = [
    "largecoeff_pisinger\\knapPI_1_10000_10000000.csv",
    "largecoeff_pisinger\\knapPI_2_10000_10000000.csv",
]
hard_instances_paths = [
    "hardinstances_pisinger\\knapPI_11_10000_1000.csv",
    "hardinstances_pisinger\\knapPI_12_10000_1000.csv",
]

# Parseo único (cada archivo produce una lista; tomamos el primer conjunto)
small_instances = [knapack_parser(Path(p))[0] for p in small_instances_paths]
large_instances = [knapack_parser(Path(p))[0] for p in large_instances_paths]
hard_instances = [knapack_parser(Path(p))[0] for p in hard_instances_paths]
```

```python
# Generación de datos (usa joblib en módulo externo)
# Ajusta seeds/tau/iteration_values si deseas menos combinaciones para prueba rápida.

# Pequeñas
generate_data(small_instances, Path("eo_small_results.csv"), n_jobs=-1, seeds=range(1, 31),
              tau_values=list(np.arange(1.0, 2.1, 0.2, dtype=float)), iteration_values=range(500, 3001, 500))
```

```python
# Grandes
generate_data(large_instances, Path("eo_large_results.csv"), n_jobs=-1, seeds=range(1, 31),
              tau_values=list(np.arange(1.0, 2.1, 0.2, dtype=float)), iteration_values=range(500, 5001, 500))
```
